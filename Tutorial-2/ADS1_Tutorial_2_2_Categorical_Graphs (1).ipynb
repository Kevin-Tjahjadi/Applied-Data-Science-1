```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# Let's start by randomly generated 4 sets of sample data sets
# Explore what is np.random.normal ( )?
sample1 = np.random.normal(-1.0, 1.0, 10000)
sample2 = np.random.normal(1.0, 0.5, 10000)
sample3 = np.random.normal(0.0, 1.5, 10000)
sample4 = np.random.normal(-0.2, 2.0, 10000)
# Test by printing out each sample data sets, Example print(sample1)
print(sample1)
print(sample2)
# Find out the data type of the samples
print(type(sample1))
```

```python
def plot_subplotted_histograms(sample1, sample2, sample3, sample4):
    """
    The following scripts will Plots 4 histograms as subplots. Understand what is subplots?
    """
    fig, axs = plt.subplots(2, 2, dpi=144)
    axs = axs.flatten()

    axs[0].hist(sample1)
    axs[1].hist(sample2)
    axs[2].hist(sample3)
    axs[3].hist(sample4)

    for i, ax in enumerate(axs):
        ax.set_ylabel('N')
        ax.set_title('Sample ' + str(i + 1))

    fig.subplots_adjust(wspace=0.5, hspace=0.5)
    plt.show()
    return
```

```python
plot_subplotted_histograms(sample1, sample2, sample3, sample4)
```

```python
def plot_overplotted_histograms(*samples):
    """
    Plots 4 histograms on top of each other
    """
    plt.figure(dpi=144)

    for i, sample in enumerate(samples):
        plt.hist(sample, label='Sample ' + str(i + 1), range=(-5, 5), bins=10, alpha=0.5)

    plt.ylabel('N')
    plt.xlim(-5, 5)
    plt.legend()
    plt.show()
    return
```

```python
plot_overplotted_histograms(sample1, sample2, sample3, sample4)
```

# Question 2

Download the following datasets from BB and upload to Google Colab

TSCO_ann, BP_ann, BCS_ann, VOD_ann

Perform the same example histograms as above, so a subplots and overplotted (as a cumulative density) version, with the financial annual returns provided.

```python
df_tesco = pd.read_csv('TSCO_ann.csv', index_col='year')
df_bp = pd.read_csv('BP_ann.csv', index_col='year')
df_barclays = pd.read_csv('BCS_ann.csv', index_col='year')
df_vodaphone = pd.read_csv('VOD_ann.csv', index_col='year')
print(df_tesco.head())
print(df_bp.head())
print(df_barclays.head())
print(df_vodaphone.head())
```

```python
print(df_tesco)
```

```python
def plot_subplotted_annual_returns(tesco, bp, barclays, vodaphone):
    """
    Plots 4 histograms of annual returns as subplots
    """
    fig, axs = plt.subplots(2, 2, dpi=144, figsize=(10, 8))
    axs = axs.flatten()

    axs[0].hist(tesco['ann_return'], color='blue', alpha=0.7)
    axs[0].set_title('Tesco Annual Returns')

    axs[1].hist(bp['ann_return'], color='orange', alpha=0.7)
    axs[1].set_title('BP Annual Returns')

    axs[2].hist(barclays['ann_return'], color='green', alpha=0.7)
    axs[2].set_title('Barclays Annual Returns')

    axs[3].hist(vodaphone['ann_return'], color='red', alpha=0.7)
    axs[3].set_title('Vodafone Annual Returns')

    for ax in axs:
        ax.set_ylabel('N')
        ax.set_xlabel('Annual Return (%)')

    fig.subplots_adjust(wspace=0.3, hspace=0.4)
    plt.show()
    return
```

```python
plot_subplotted_annual_returns(df_tesco, df_bp, df_barclays, df_vodaphone)
```

```python
def plot_overplotted_annual_returns(tesco, bp, barclays, vodaphone):
    """
    Plots 4 histograms on top of each other as a cumulative density version
    """
    plt.figure(dpi=144, figsize=(8, 5))

    # Using density=True and cumulative=True to match "cumulative density"
    # Removing range=(-5, 5) restriction to fit the actual financial returns
    plt.hist(tesco['ann_return'], label='Tesco', bins=20, alpha=0.5, density=True, cumulative=True)
    plt.hist(bp['ann_return'], label='BP', bins=20, alpha=0.5, density=True, cumulative=True)
    plt.hist(barclays['ann_return'], label='Barclays', bins=20, alpha=0.5, density=True, cumulative=True)
    plt.hist(vodaphone['ann_return'], label='Vodafone', bins=20, alpha=0.5, density=True, cumulative=True)

    plt.xlabel('Annual Return (%)')
    plt.ylabel('Cumulative Density')
    plt.title('Cumulative Density of Annual Returns')
    plt.legend()
    plt.show()
    return
```

```python
plot_overplotted_annual_returns(df_tesco, df_bp, df_barclays, df_vodaphone)
```

Let's finish with some pie charts.
For example, the largest countries in the EU, by GDP.

```python
gdp = np.array([3132.670e9, 2225.260e9, 1672.438e9, 1113.851e9])
countries = ["Germany", "France", "Italy", "Spain"]
```

```python
def plot_gdp_pie(gdp, labels=countries):
    """
    Creates a pie chart of the GDPs of the 4 largest countries, by GDP
    """
    plt.figure(dpi=144)

    plt.pie(gdp, labels=labels, autopct='%1.1f%%')
    plt.title('Largest Economies in the EU')
    plt.axis('Equal')
    plt.show()
    return
```

```python
plot_gdp_pie(gdp)
# but this isn't all of the economies in the EU!
```

```python
gdp = np.array([3132.670e9, 2225.260e9, 1672.438e9, 1113.851e9])
countries = ["Germany", "France", "Italy", "Spain"]

gdp_EU = 12451.987e9
gdp /= gdp_EU
```

```python
def plot_gdp_pie_unnormalised(gdp, labels=countries):
    """
    Creates a pie chart of the GDPs of the 4 largest countries, by GDP
    """
    plt.figure(dpi=144)

    plt.pie(gdp, labels=labels, autopct='%1.1f%%', normalize=False)
    plt.title('Largest Economies in the EU')
    plt.axis('Equal')
    plt.show()
    return
```

```python
plot_gdp_pie_unnormalised(gdp)
```

# Question 3
Create these same style of pie charts, using the known market caps of the previously used companies, and relate it to the entire FTSE cap.
<ul>
    <li>Tesco: 20,979</li>
    <li>BP: 68,785</li>
    <li>Barclays: 33,367</li>
    <li>Vodaphone: 29,741</li>
    <li>FTSE: 1,814,000</li>
</ul>

You are to plot 2 Pie-Charts:
 1) Individual companies share  
 2) Total market cap of 4 companies

```python
cap = np.array([20979, 68785, 33367, 29741])
companies = ['Tesco', 'BP', 'Barclays', 'Vodafone']
ftse = 1814000
norm_cap = cap / ftse
```

```python
def plot_market_cap_pie(cap, labels=companies):
    """
    Creates a pie chart of the market cap of 4 major companies
    """
    plt.figure(dpi=144)
    plt.pie(cap, labels=labels, autopct='%1.1f%%')
    plt.title('Individual Market Share Among the 4 Companies')
    plt.axis('equal')
    plt.show()
    return
```

```python
plot_market_cap_pie(cap)
```

```python
def plot_market_cap_pie_unnormalised(norm_cap, labels=companies):
    """
    Creates a pie chart of the market cap of 4 major companies, normalized to the FTSE
    """
    plt.figure(dpi=144)
    # Include the rest of the FTSE cap to complete the 100%
    all_caps = np.append(norm_cap, 1.0 - np.sum(norm_cap))
    all_labels = labels + ['Other FTSE']

    plt.pie(all_caps, labels=all_labels, autopct='%1.2f%%', normalize=True)
    plt.title('Market Cap of 4 Companies Relative to FTSE')
    plt.axis('equal')
    plt.show()
    return
```

```python
plot_market_cap_pie_unnormalised(norm_cap)
```
